```python
# 📚 StudyFlow — Agentic AI Study Planner
```

```python
from memory import AgentMemory
from tools import StudyTools
from agent import StudyFlowAgent

memory = AgentMemory()
tools = StudyTools(memory)

study_agent = StudyFlowAgent(
    memory=memory,
    tools=tools
)

print("✅ StudyFlow initialized successfully!")
```

```python
result = study_agent.run(
    "Who invented the telephone?"
)

print(result["response"])
```

```python
result = study_agent.run(
    "I have a DBMS normalization assignment due on September 8, 2026. "
    "It will take 4 hours. I can study 3 hours per day. "
    "Make me a study plan."
)

print(result["response"])
```

```python
print("=" * 60)
print("              🤖 AGENT TRACE")
print("=" * 60)

for step in result["trace"]:
    print()
    print(f"STEP {step['step']}")
    print(f"TOOL      : {step['tool']}")
    print(f"ARGUMENTS : {step['arguments']}")
    print(f"RESULT    : {step['result']}")
    print("-" * 60)
```
