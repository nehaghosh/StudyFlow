```python
import importlib
import tools

importlib.reload(tools)

from tools import StudyTools
from memory import AgentMemory

memory = AgentMemory()
tools = StudyTools(memory)

print("StudyFlow tools loaded successfully!")
```

```python
tools.add_task(
    "DBMS",
    "Complete normalization assignment",
    "2026-09-08",
    4
)

tools.add_task(
    "AI",
    "Complete StudyFlow project",
    "2026-09-10",
    6
)

print("Tasks added successfully!")
```

```python
schedule = tools.build_schedule(
    daily_hours=3,
    start_date="2026-09-04"
)

print(schedule)
```

```python
print("=" * 60)
print("              📚 STUDYFLOW STUDY PLAN")
print("=" * 60)

if schedule["success"]:

    for item in schedule["schedule"]:

        print()
        print("📅 Date     :", item["date"])
        print("📖 Subject  :", item["subject"])
        print("📝 Task     :", item["task"])
        print("⏱️ Study    :", item["hours"], "hour(s)")
        print("🎯 Deadline :", item["deadline"])
        print("-" * 60)

    if schedule["unscheduled"]:

        print()
        print("⚠️ WARNING — NOT ENOUGH TIME")
        print("=" * 60)

        for item in schedule["unscheduled"]:

            print()
            print("📖 Subject :", item["subject"])
            print("📝 Task    :", item["task"])
            print("⏳ Remaining:", item["remaining_hours"], "hour(s)")
            print("🎯 Deadline :", item["deadline"])

    else:

        print()
        print("✅ ALL TASKS CAN BE COMPLETED!")
        print("=" * 60)

else:

    print("❌ ERROR:", schedule["error"])
```

```python
urgent_schedule = tools.build_schedule(
    daily_hours=3,
    start_date="2026-09-04",
    custom_hours={
        "2026-09-04": 6,
        "2026-09-05": 5,
        "2026-09-06": 2,
        "2026-09-07": 4
    }
)

print("=" * 60)
print("                 🔥 CUSTOMIZED STUDY PLAN")
print("=" * 60)

for item in urgent_schedule["schedule"]:

    print()
    print("📅 Date     :", item["date"])
    print("📖 Subject  :", item["subject"])
    print("📝 Task     :", item["task"])
    print("⏱️ Study    :", item["hours"], "hour(s)")
    print("🎯 Deadline :", item["deadline"])
    print("-" * 60)

if urgent_schedule["unscheduled"]:
    print("\n⚠️ Some work could not fit before the deadline.")
else:
    print("\n✅ ALL TASKS CAN BE COMPLETED!")
```
